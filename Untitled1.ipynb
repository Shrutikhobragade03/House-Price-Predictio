```python
#Print numerical columns whose correlation with SalePrice is between 0.2 and 0.4
for col in df.select_dtypes(exclude='object').columns:
    
    corr = df[col].corr(df['SalePrice'])
    
    if 0.2 <= corr <= 0.4:
        print(col, ":", round(corr, 3)) # corr,3 means correlation values rounded to  3 decimal places
        
```

## Mean absolute error 
* actual - predicted 
* aaise krke sabhi values ka difference collect krke uska average nikalta hai aur usko MSE kehte haii


## Mean squared error 
* (actual - predicted)** = error ka square
* why to square?
* to punish large mistakes more heavily
* a huge error becomes more important

## Root mean squared error
RMSE = √MSE
* Why RMSE?
* Because MSE is in squared units, which are hard to interpret.
* RMSE converts it back to the original unit.
